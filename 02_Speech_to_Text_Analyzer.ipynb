```python
# ============================================================
# APPLICATION 2
# SPEECH-TO-TEXT ANALYZER
# Text & Speech Analysis
# Google Colab
# ============================================================

# Install required libraries
!pip -q install SpeechRecognition pydub wordcloud

import speech_recognition as sr
from google.colab import files
from IPython.display import display, HTML
from collections import Counter
import re
import wave
import os


# ============================================================
# APPLICATION HEADER
# ============================================================

display(HTML("""
<div style="
background:#f5f7fb;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
">

<h1>Speech-to-Text Analyzer</h1>

<p style="font-size:18px;">
Convert speech into text and analyze the content
</p>

<p>
Text & Speech Analysis - Application 2
</p>

</div>
"""))


# ============================================================
# UPLOAD AUDIO
# ============================================================

print("Upload your audio file.")
print("Recommended format: WAV")

uploaded = files.upload()

audio_file = list(uploaded.keys())[0]

print("\nUploaded file:")
print(audio_file)


# ============================================================
# SPEECH RECOGNITION
# ============================================================

recognizer = sr.Recognizer()

print("\nProcessing your speech...")
print("Please wait...")


try:

    with sr.AudioFile(audio_file) as source:

        audio = recognizer.record(source)

    text = recognizer.recognize_google(audio)

except sr.UnknownValueError:

    text = ""

    print("\nCould not understand the speech.")

except sr.RequestError:

    text = ""

    print("\nSpeech recognition service is unavailable.")


# ============================================================
# DISPLAY TRANSCRIBED TEXT
# ============================================================

print("\n" + "=" * 60)

print("TRANSCRIBED TEXT")

print("=" * 60)

if text:

    print("\n" + text)

else:

    print("\nNo text was detected.")


# ============================================================
# TEXT ANALYSIS
# ============================================================

if text:

    # Words
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    # Sentences
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    # Characters
    characters = len(text)

    # Word count
    word_count = len(words)

    # Sentence count
    sentence_count = len(sentences)

    # Average word length
    if word_count > 0:
        average_word_length = sum(
            len(word) for word in words
        ) / word_count
    else:
        average_word_length = 0

    # Most common words
    common_words = Counter(words).most_common(10)


    # ========================================================
    # DISPLAY ANALYSIS
    # ========================================================

    print("\n" + "=" * 60)

    print("TEXT ANALYSIS")

    print("=" * 60)

    print(f"\nCharacters       : {characters}")

    print(f"Words            : {word_count}")

    print(f"Sentences        : {sentence_count}")

    print(
        f"Average word length : "
        f"{average_word_length:.2f}"
    )


    # ========================================================
    # MOST FREQUENT WORDS
    # ========================================================

    print("\nMost Frequent Words")

    print("-" * 40)

    for word, count in common_words:

        print(f"{word:<20} {count}")


    # ========================================================
    # VISUAL DASHBOARD
    # ========================================================

    rows = ""

    for word, count in common_words:

        rows += f"""
        <tr>
            <td>{word}</td>
            <td>{count}</td>
        </tr>
        """


    display(HTML(f"""

    <div style="
    max-width:800px;
    margin:30px auto;
    font-family:Arial;
    ">

    <div style="
    background:white;
    padding:25px;
    border-radius:18px;
    border:1px solid #ddd;
    ">

    <h2 style="text-align:center;">
    Speech Analysis Report
    </h2>

    <hr>

    <h3>Transcribed Speech</h3>

    <div style="
    background:#f5f7fb;
    padding:18px;
    border-radius:10px;
    line-height:1.6;
    ">

    {text}

    </div>

    <br>

    <h3>Statistics</h3>

    <table style="
    width:100%;
    border-collapse:collapse;
    text-align:center;
    ">

    <tr>
        <th style="padding:10px;border:1px solid #ddd;">
        Metric
        </th>

        <th style="padding:10px;border:1px solid #ddd;">
        Value
        </th>
    </tr>

    <tr>
        <td style="padding:10px;border:1px solid #ddd;">
        Characters
        </td>

        <td style="padding:10px;border:1px solid #ddd;">
        {characters}
        </td>
    </tr>

    <tr>
        <td style="padding:10px;border:1px solid #ddd;">
        Words
        </td>

        <td style="padding:10px;border:1px solid #ddd;">
        {word_count}
        </td>
    </tr>

    <tr>
        <td style="padding:10px;border:1px solid #ddd;">
        Sentences
        </td>

        <td style="padding:10px;border:1px solid #ddd;">
        {sentence_count}
        </td>
    </tr>

    <tr>
        <td style="padding:10px;border:1px solid #ddd;">
        Average Word Length
        </td>

        <td style="padding:10px;border:1px solid #ddd;">
        {average_word_length:.2f}
        </td>
    </tr>

    </table>

    <br>

    <h3>Most Frequent Words</h3>

    <table style="
    width:100%;
    border-collapse:collapse;
    text-align:center;
    ">

    <tr>
        <th style="padding:10px;border:1px solid #ddd;">
        Word
        </th>

        <th style="padding:10px;border:1px solid #ddd;">
        Frequency
        </th>
    </tr>

    {rows}

    </table>

    </div>

    </div>

    """))

else:

    print("\nPlease upload a clear speech recording and try again.")
```
